#### Advanced Indexing and Slicing

###### Create a 5×5 NumPy array with random integers between 1 and 100.

```python
array1 = np.random.randint(1,101,size = (5,5))
print(array1)
```

###### Replace all values greater than the column mean with -1.

```python
col_mean = np.mean(array1,axis = 0)
print(col_mean)
```

```python
mask = np.where(array1>col_mean)
```

```python
array1[mask]=-1
```

#### Real-World Normalization Task

###### Simulate a dataset with 12 samples and 4 features.

```python
data  = np.random.randint(1,100,size = (12,4))
data
```

###### Perform : Min- Max normalization , z-score normalization

```python
min_data = np.min(data , axis = 0)
max_data = np.max(data , axis = 0)

nor = (data - min_data) / (max_data - min_data)
nor
```

```python
mean_data = np.mean(data , axis = 0)
std_data = np.std(data , axis = 0)

nor_z = (data - mean_data )/ std_data
nor_z
```
