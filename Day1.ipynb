#### Array Manipulation and Slicing

###### Create a 5×5 matrix of random integers between 1–50.

```python
array = np.random.randint(1, 51, size = (5,5))
array
```

```python
#Basic approach
def is_prime_basic(n):
    if n<= 1:
        return False
    for i in range(2,n):
        if n%i == 0:
            return False
    return True

for  i in range(array.shape[0]):
    for j in range(array.shape[1]):
        if is_prime_basic(array[i,j]):
            array[i,j]=0
array  
```

```python
#n to sqrt of n and using np.ndenumerate()
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(np.sqrt(n))+1):
        if n%i == 0:
           return False
    return True

for indx , val in np.ndenumerate(array):
    if is_prime(val):
        array[indx]=0
array
```

```python
#vectorized
def is_prime_vec(n):
    if n <= 1:
        return False
    return all(n % i != 0 for i in range(2, int(np.sqrt(n))+1))
prime_mask = np.vectorize(is_prime_vec)(array)
array[prime_mask] = 0

array
```

###### Extract the diagonal elements and calculate their sum.

```python
#Basic approach
sum=0
for i in range(array.shape[0]):
    sum += array[i,i]
sum
```

```python
#using inbuilt functions
diag_ele = np.diag(array)
print(diag_ele)

sum_diag_ele  = np.sum(diag_ele)
print(sum_diag_ele)

#trace inbuilt function
print(np.trace(array))
```

#### Real-World Data Simulation

###### Simulate a dataset with 10 samples and 4 features using random numbers.

```python
#10 samples -> observation(rows)
#4 features -> variables(columns)
data  = np.random.rand(10,4)
print(data)
```

###### Standardize the dataset (mean=0, std=1) without using sklearn.

```python
print("=====Actual Data======")
print(data)
mean = np.mean(data, axis = 0) #mean of each feature
std = np.std(data, axis = 0)  #std of each feature

stand_data = (data - mean) / std
print("=====Standardized Data=======")
print(stand_data)
```

```python
#verification whether the data is standardized or not
print(np.mean(stand_data, axis = 0))   #mean = 0
print(np.std(stand_data, axis = 0))  #std = 1
```

```python
print(np.round(np.mean(stand_data, axis = 0)))
```
